## Introduction to creating your own circuits.

Here the cells below can be copied and you can try out different circuits. Consider trying to implement a circuit that produces a 3-qubit circuit that produces a Greenberge-Horne-Zeilinger (GHZ) state. This is an  entangled quantum state that involves at least three qubits, which is named for the three authors that first described this state. The GHZ state predicts outcomes from experiments that directly contradict predictions by every classical local hidden-variable theory. 

The GHZ state can be written as: 
$
\ket{\mathrm {GHZ}} =\frac {\ket{000}+\ket{111}}{\sqrt {2}}.
$

For more information see [here](https://en.wikipedia.org/wiki/Greenberger%E2%80%93Horne%E2%80%93Zeilinger_state).


```python
# sandbox to play with circuits

# now boiler plate code for CNOT
dev = qml.device("default.qubit", wires=["|q1>", "|q2>", "|q3>",])

# now simple circiut where we return the state of the qubits
@qml.qnode(dev)
def circuit():
    qml.Hadamard(wires="|q1>")
    qml.CNOT(["|q1>","|q2>"])
    return qml.counts(all_outcomes=True)
# what does your circuit look like 
fig, ax = qml.draw_mpl(circuit, show_all_wires=True)()
fig.show()

# lets create a quantum device that will take 100 shots
results = circuit(shots=100)
# here plot the results
PlotPennyLaneHisto(results, 'Your experiment')
```
